```python
# WHAT IS THE COUNT OF PURCHASED AND NOT PURCHASED IN "WITH COVER" CATREGORY & 
# WHAT IS THE COUNT OF PURCHASED AND NOT PURCHASED IN "WITH PHONE" CATREGORY - FORMS CONTINGENCY TABLE 
# THIS IS WHAT WE ARE FINDING USING CROSSTAB

CT_Values = pd.crosstab(Recom_data ["SUGGESTION_TYPE"], Recom_data["PURCHASE_FLAG"]).values
CT_Values
```

```python
#COMPUTE PURCHASE RATE
With_Phone_Rate = 153 / (153+200)
With_Cover_Rate = 195 / (195+165)
print("Phone With ScreenGuard", With_Phone_Rate)
print("Cover With ScreenGuard", With_Cover_Rate)
```

NULL HYPOTHESIS = "THERE IS NO RELATIONSHIP B/W PURCHASE RATE & RECOMMENDATION TYPE. THEY ARE INDEPENDENT"
ALTERNATE HYPOTHESIS = " THERE IS RELATIONSHIP B/W PURCHASE RATE & RECOMMENDATION TYPE. THEY ARE NOT INDEPENDENT"

```python
from scipy.stats import chi2_contingency
chi2_statistic, p_value, dof, expected_values = chi2_contingency(CT_Values, correction = False)
print(chi2_statistic, p_value)
```

```python
prob = 0.95
alpha = 1.0 - prob
print('significant=%.3f, p=%.3f' % (alpha, p_value))
if p_value <= alpha:
    print('Dependent, we can reject H0 safely')
else:
    print('Independent, No Sufficient Evidence to Reject H0')
    
```
